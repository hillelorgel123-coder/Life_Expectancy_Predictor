```python
import pandas as pd

# Define the Gompertz-Makeham function
def calculate_qx(age, A, B, c):
    return A + B * (c ** age)

# Set up the DataFrame for ages 18 to 120
ages = list(range(18, 121))
df = pd.DataFrame({'Age': ages})

# Calculate baseline probabilities using your constants
df['Male_qx'] = df['Age'].apply(lambda x: calculate_qx(x, 0.0003, 0.00004, 1.095))
df['Female_qx'] = df['Age'].apply(lambda x: calculate_qx(x, 0.0002, 0.00002, 1.095))

# Display the benchmark ages to verify the math
benchmarks = df[df['Age'].isin([25, 45, 65, 85])]
print(benchmarks)
```

```python
# Placeholder rates representing the official SOA 2015 VBT
vbt_data = {
    'Age': [25, 45, 65, 85],
    'VBT_Male_qx': [0.000650, 0.002600, 0.015000, 0.090000], 
    'VBT_Female_qx': [0.000350, 0.001400, 0.007500, 0.045000] 
}

# Convert to a pandas DataFrame and merge with your generated benchmarks
vbt_df = pd.DataFrame(vbt_data)
comparison = pd.merge(benchmarks, vbt_df, on='Age')

# Calculate the absolute difference (residuals) between your model and the VBT
comparison['Male_Diff'] = comparison['Male_qx'] - comparison['VBT_Male_qx']
comparison['Female_Diff'] = comparison['Female_qx'] - comparison['VBT_Female_qx']

# Print the final comparison table
print(comparison[['Age', 'Male_qx', 'VBT_Male_qx', 'Male_Diff', 'Female_qx', 'VBT_Female_qx', 'Female_Diff']])
```
